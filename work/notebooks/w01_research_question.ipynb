## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The decision this project aims to improve is **which content pages should be reviewed first**.

The primary users are content editors or SEO specialists. They would use the ranked recommendations to decide which pages should be refreshed, expanded, monitored, or investigated further.

The output of the project will be a ranked list of content pages, together with a priority score and supporting reason codes explaining why each page was recommended.

A false positive would recommend a page that does not actually require attention, wasting valuable review time and resources. A false negative would fail to identify a page that could have benefited from review, causing a potentially valuable opportunity for improvement to be missed.

Machine learning is appropriate because page performance depends on many interacting signals, such as traffic, engagement, content characteristics, and search performance. These relationships are too complex to capture reliably with a single hand-written rule.
