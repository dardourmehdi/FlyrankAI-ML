## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is Refresh / Content Opportunity Scoring, framed as a binary classification task.

The model would predict whether a content item is declining or not declining. Although the target is binary, the model would produce a probability for each content item. This probability can be used as a priority score to rank the pages that should be reviewed first.and

Classification is therefore the underlying ML task, while scoring and ranking are how the output supports the content decision.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target us **is_declining_labal**.

A value of **1** means that the observed **trend_direction** is down. In the contrary, a value of **0** means that the observed trend direction is not down.

This label is a defined proxy, created from a measured trend rather than from a direct observation that a page truly needs to be refreshed. Decline is treated as a signal that a page may deserve review, but it does not prove that refreshing the page is the correct action.

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")
df["is_declining_label"] = (df["trend_direction"].astype(str).str.lower() == "down").astype(int)
df["is_declining_label"].value_counts().sort_index()
```

```python
df["is_declining_label"].value_counts(normalize=True).sort_index() * 100 #to have the percentage 
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The succes metric used here is Precision@50.

It measures the proportion of declining content items among the first 50 items recommended for review. This matches the real decision because a content specialist has limited review time and will mainly act on the top recs.

The observed prevalence of the proxy label is 54.2%, so a non-informative ranking would eontain approximately 27 declining items in its top 50.

```python
declining_rate = df["is_declining_label"].mean()
expected_top_50 = declining_rate * 50

print(f"Observed declining rate : {declining_rate:.1%}")
print("Expected declining items in a non-informative top 50:", round(expected_top_50, 1))
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule is enough to create the proxy label: if trend_direction is "down", the content item is labelled as declining.

However, this rule does not show which declining pages represent the strongest review opportunities. Two pages may both be declining, but one may be more important because of its traffic, previous performance or search visibility.

A ML model could combine these different signals to produce a decision-support score and prioritize the pages that should be reviewed first. The model would not prove that refreshing a page will improve its performance, but it could help make the review process more focused.
